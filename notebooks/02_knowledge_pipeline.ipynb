```python
import os
import sys
from pathlib import Path
import pandas as pd
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

# Anchor project root
project_root = Path(r"C:\Users\DELL\dsba6171-microcosm-starter")
os.chdir(project_root)

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.data.loader import load_and_split_document

# Define paths
csv_path = project_root / "data" / "structured" / "raw" / "synthetic_knowledge_items.csv"
chroma_dir = project_root / "data" / "processed" / "chroma"

# 1. Ensure CSV contains the expected columns 'ki_topic' and 'ki_text'
sample_data = pd.DataFrame([
    {
        "ki_id": "KI-001",
        "ki_topic": "ChromaDB Vectorstore",
        "ki_text": "ChromaDB is an open-source vector database designed to store embeddings and enable efficient similarity search for RAG systems."
    },
    {
        "ki_id": "KI-002",
        "ki_topic": "LangChain Framework",
        "ki_text": "LangChain provides components for loading data, generating text embeddings, and chaining vector retrievers with LLMs."
    },
    {
        "ki_id": "KI-003",
        "ki_topic": "Qwen Model Integration",
        "ki_text": "Qwen2.5-0.5B-Instruct is a lightweight instruction-tuned model used to synthesize answers based on retrieved context."
    }
])
sample_data.to_csv(csv_path, index=False)

# 2. Load and split documents
documents = load_and_split_document(str(csv_path))

# 3. Create and persist ChromaDB vectorstore
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    persist_directory=str(chroma_dir)
)

print(f"Processed {len(documents)} document chunks.")
print(f"Vectorstore initialized and persisted successfully at: {chroma_dir}")
```
