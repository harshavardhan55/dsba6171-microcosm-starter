```python
import os
import sys
from pathlib import Path
import pandas as pd
from langchain_community.document_loaders import CSVLoader

# 1. Locate project root by searching for 'src' starting from the notebook file location or working directory
def find_project_root():
    # Try using __file__ if available, otherwise search around current working path
    search_paths = [Path.cwd()]
    
    # Walk parent directories for any path containing 'src'
    for base in search_paths:
        for p in [base] + list(base.parents):
            if (p / "src").is_dir():
                return p
                
    # If working dir was moved to C:\, check user directory or typical project locations
    for p in Path.home().rglob("dsba6171-microcosm-starter"):
        if (p / "src").is_dir():
            return p
            
    raise FileNotFoundError("Could not locate repository root containing 'src'.")

project_root = find_project_root()
os.chdir(project_root)

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

# 2. Define absolute paths within the repository
raw_data_dir = project_root / "data" / "structured" / "raw"
csv_path = raw_data_dir / "synthetic_knowledge_items.csv"

raw_data_dir.mkdir(parents=True, exist_ok=True)

# 3. Create dataset
sample_data = pd.DataFrame([
    {
        "ki_id": "KI-001",
        "title": "ChromaDB Vectorstore",
        "content": "ChromaDB is an open-source vector database designed to store embeddings and enable efficient similarity search for RAG systems."
    },
    {
        "ki_id": "KI-002",
        "title": "LangChain Framework",
        "content": "LangChain provides components for loading data, generating text embeddings, and chaining vector retrievers with LLMs."
    },
    {
        "ki_id": "KI-003",
        "title": "Qwen Model Integration",
        "content": "Qwen2.5-0.5B-Instruct is a lightweight instruction-tuned model used to synthesize answers based on retrieved context."
    }
])

sample_data.to_csv(csv_path, index=False)

# 4. Load documents using CSVLoader
loader = CSVLoader(file_path=str(csv_path), metadata_columns=["ki_id"])
chunks = loader.load()

print(f"Project Root correctly set to: {project_root}")
print(f"Dataset saved at: {csv_path}")
print(f"Total chunks created: {len(chunks)}")
print("\nSample Chunk Metadata:", chunks[0].metadata)
print("\nSample Chunk Content:\n", chunks[0].page_content)
```
